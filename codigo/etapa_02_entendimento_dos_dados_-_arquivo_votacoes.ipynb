Dado que a quantidade de votaões é maior do que a quantidade de votações, há proposições que são votadas mais do que uma vez.

```python
votacoes_da_mesma_proposicao = df_votacoes.groupby('ultimaApresentacaoProposicao_idProposicao').filter(lambda x: x['id'].count() >= 2)
id_proposicao_com_mais_do_que_uma_votacao = votacoes_da_mesma_proposicao.iloc[1]['ultimaApresentacaoProposicao_idProposicao']
df_votacoes[df_votacoes['ultimaApresentacaoProposicao_idProposicao'] == id_proposicao_com_mais_do_que_uma_votacao]
```
