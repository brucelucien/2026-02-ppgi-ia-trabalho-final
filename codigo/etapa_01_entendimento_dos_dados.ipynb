# Significado das variáveis nos microdados

```python
def gerar_texto_significado_variavel(campo):
    texto = ''
    texto += NOVA_LINHA
    texto += NOVA_LINHA + "===================="
    texto += NOVA_LINHA + "NOME:"
    texto += NOVA_LINHA + str(campo["Nome da Variável"])
    texto += NOVA_LINHA
    texto += NOVA_LINHA + "DESCRIÇÃO:"
    texto += NOVA_LINHA + str(campo["Descrição da Variável"])
    texto += NOVA_LINHA
    texto += NOVA_LINHA + "CATEGORIA:"
    texto += NOVA_LINHA + str(campo["Categoria"])
    texto += NOVA_LINHA
    texto += NOVA_LINHA + "TIPO:"
    texto += NOVA_LINHA + str(campo["Tipo"])
    texto += NOVA_LINHA
    texto += NOVA_LINHA + "TAMANHO:"
    texto += NOVA_LINHA + str(campo["Tamanho"])
    texto += NOVA_LINHA
    texto += NOVA_LINHA + "OBSERVAÇÕES:"
    texto += NOVA_LINHA + str(campo["Observações"])
    return str(texto)

def gerar_texto_significado_todas_variaveis(df):
    texto = ''
    for variavel in df.iloc:
        texto += gerar_texto_significado_variavel(variavel)
    return texto
```

```python
print(gerar_texto_significado_todas_variaveis(df_dicionario))
```

# Registrando significado das variáveis no diretório de processamento de dados

```python
with open(ENDERECO_GERACAO_RESUMO_SIGNIFICADO_VARIAVEIS, 'w') as arquivo:
    arquivo.write(gerar_texto_significado_todas_variaveis(df_dicionario))
```
