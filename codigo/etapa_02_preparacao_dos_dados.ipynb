#Substituição de dados  = -1, 9, 99 2  conforme a características dos dados:
#-1 represnta um salto no quetionário em questões que dependem da resposta de questões anteriores
#-2 representa uma desistência de responder ao questionário
#9/99 representam valores não informados, geralmente um tratamento de dado vazio



```python
#%pip install pandas

import os
import pandas as pd
import numpy as np

# 1. Definir os caminhos dos arquivos e diretórios
caminho_entrada = "../dados/em_processamento/dados_preparados.txt"
diretorio_saida = "../dados/em_processamento/"
#caminho_saida = "dados/dados_preparados_v2.txt"
caminho_saida = os.path.join(diretorio_saida, "dados_preparados_v2.txt")

# Garantir que o diretório de saída existe
os.makedirs(diretorio_saida, exist_ok=True)

# 2. Ler os dados para o dataframe 'dados_origem' (separado por ponto e vírgula)
# Utiliza low_memory=False para evitar avisos caso haja colunas mistas
dados_origem = pd.read_csv(caminho_entrada, sep=";", low_memory=False)

# 3. Copiar os dados para o novo dataframe 'dados_normalizados'
dados_normalizados = dados_origem.copy()

# 4. Mapeamento das regras de conversão (-1 para o valor específico de cada variável)
regras_conversao = {
    # Bloco B04
    "B04002A": 0,
    "B04001": 0,
    "B04003": 1,
    "B04006B": 1,
    #"B04009B": 0,
    "B04012": 2,
    "B04013": 2,
    "B04014A": 2,
    "B04015A": 2,
    "B04011B01": 0,
    "B04011B02": 0,
    "B04011B03": 0,
    "B04011B04": 0,
    "B04011B05": 0,
    "B04011B06": 0,
    # Bloco B05
    "B06001A": 0,
    "B05002A": 0,
    "B05003A": 0,
    "B05007A": 0,
    "B05009A": 1,
    "B05004A": 1,
    "B05005A": 0,
    #"B05006C": 0,    
    "B05011": 0,
    # Bloco B06
    "B06006A": 0,
    "B0600701": 0,
    "B0600702": 0,
    "B0600703": 0,
    "B0600704": 0,
    "B06002A": 0,
    "B06008": 0,
    "B06003B": 1,
    "B06004A": 1,
    "B06009": 1,
    # Bloco B07 (Parte 1)
    "B07008A01": 0,
    "B07008A02": 0,
    "B07008A03": 0,
    "B07008A04": 0,
    "B07008A05": 0,
    "B07008A06": 0,
    "B07008A07": 0,
    "B07008A08": 0,
    "B07008A09": 0,
    "B07008A10": 0,
    # Bloco B07 (Parte 2)
    "B0701401": 0,
    "B0701402": 0,
    "B0701403": 0,
    "B0701404": 0, 
    "B0701405": 0,
    "B0701406": 0,
    "B0701407": 0,
    "B0701408": 0,
    "B0701409": 0,
    "B0701410": 0,
    # Bloco B09 (Parte 1)
    "B09014B": 0,
    "B09010B201": 0,
    "B09010B202": 0,
    "B09010B203": 0,
    "B09010B204": 0,
    "B09010B205": 0,
    "B09010B206": 0,
    "B09010B207": 0,
    "B09016A1": 0,
    "B09017B101": 0,
    "B09017B102": 0,
    "B09017B103": 0,
    "B09017B104": 0,
    "B09017B105": 0,
    "B09017B106": 0,
    # Bloco B09 (Parte 2)
    "B09016A2": 0,
    "B09017B201": 0,
    "B09017B202": 0,
    "B09017B203": 0,
    "B09017B204": 0,
    "B09017B205": 0,
    "B09017B206": 0,
    "B09018": 0,
    # Bloco B13
    "B13002B": 0,
    "B13003A": 0,
    #"B13004B": 0,
    "B13007A": 0,
    #target
    "B12004": 0,"B12005": 0,"B12006": 0,"B12007": 0,"B12008": 0,"B12009": 0
}
#corrigir código para substiuir  qq valor alternativo das targets  por 0
# 5. Aplicar as conversões verificando se a coluna existe no dataframe
for coluna, valor_substituto in regras_conversao.items():
  if coluna in dados_normalizados.columns:
    # Substitui -1 pelo valor correspondente especificado
    dados_normalizados[coluna] = dados_normalizados[coluna].replace(
        -1, valor_substituto
    )
    if(coluna in VARIAVEIS_DEPENDENTES):
        dados_normalizados[coluna] = dados_normalizados[coluna].replace({99: valor_substituto, -2: valor_substituto, -9: valor_substituto})
     
  else:
    print(
        f"Aviso: A coluna '{coluna}' especificada nas regras não foi encontrada"
        " no arquivo."
    )

#6. Substitui simultaneamente -99, -2 e -9 por NaN em todo o dataframe 'dados_normalizados'
dados_normalizados = dados_normalizados.replace({-99: np.nan, -2: np.nan, -9: np.nan})
print(
    "Valores -99, -2 e -9 substituídos por NaN com sucesso."
)
```

```python

# 7. VERIFICAÇÃO PÓS-SUBSTITUIÇÃO

valores_procurados = [-1, -2, -9, -99]
colunas_com_pendencias = {}

for col in dados_normalizados.columns:
  # Verifica se a coluna possui algum dos valores restantes
  # Usamos isin para verificar se o valor existe na série (ignorando NaNs automaticamente)
  encontrados = dados_normalizados[col][
      dados_normalizados[col].isin(valores_procurados)
  ].unique()
  if len(encontrados) > 0:
    colunas_com_pendencias[col] = list(encontrados)

print("\n--- RELATÓRIO DE VERIFICAÇÃO PÓS-SUBSTITUIÇÃO ---")
if colunas_com_pendencias:
  print(
      "Atenção: As seguintes colunas ainda contêm valores (-1, -2, -9 ou -99):"
  )
  for coluna, vals in colunas_com_pendencias.items():
    print(f"  - Coluna '{coluna}': contém os valores {vals}")
else:
  print(
      "Sucesso! Nenhuma coluna no dataframe contém os valores -1, -2, -9 ou -99."
  )
print("-------------------------------------------------\n")
# Salvar o dataframe normalizado no arquivo de saída
#dados_normalizados.to_csv(caminho_saida, sep=";", index=False)
#print(f"Arquivo final salvo com sucesso em: '{caminho_saida}'.")
```

```python

# 8. FILTRAGEM DE LINHAS (Remover IDADE_AGREG == 1)

if "IDADE_AGREG" in dados_normalizados.columns:
  total_antes = len(dados_normalizados)

  # Mantém apenas as linhas onde IDADE_AGREG é diferente de 1 (ou aceita se for string/numérico)
  dados_normalizados = dados_normalizados[
      dados_normalizados["IDADE_AGREG"] != 1
  ]

  total_depois = len(dados_normalizados)
  removidas = total_antes - total_depois
  print(
      f"Filtro aplicado: {removidas} linhas removidas onde IDADE_AGREG = 1."
      f" Total de linhas restantes: {total_depois}"
  )
else:
  print(
      "Aviso: A coluna 'IDADE_AGREG' não foi encontrada no dataframe para"
      " aplicar o filtro."
  )

# 9. Salvar o dataframe normalizado e filtrado no arquivo de saída
dados_normalizados.to_csv(caminho_saida, sep=";", index=False)
print(f"Arquivo final atualizado e salvo em: '{caminho_saida}'.")
```
